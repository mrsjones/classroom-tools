```python
# ── Imports & helper functions — run once, then collapse ─────────────────────
from ColabTurtlePlus.Turtle import *
import random, math

CANVAS_WIDTH  = 800
CANVAS_HEIGHT = 500

# ── Coordinate conversion ─────────────────────────────────────────────────────
# ColabTurtlePlus uses screen coordinates: (0,0) = top-left, y increases downward.
# All helper functions accept CENTERED math coordinates: (0,0) = center,
# y increases upward (same as the original Tynker project).
# _goto() converts internally so you never have to think about screen coords.

CX = CANVAS_WIDTH  // 2   # 400 — horizontal center in screen pixels
CY = CANVAS_HEIGHT // 2   # 250 — vertical   center in screen pixels

def _goto(x, y):
    """Convert centered math coords → ColabTurtlePlus screen coords, then move."""
    goto(CX + x, CY - y)

def init_scene():
    """Initialize the canvas. Always call first inside draw_landscape()."""
    initializeTurtle(initial_speed=13)
    hideturtle()
    penup()


# ── PRIMITIVE SHAPES ─────────────────────────────────────────────────────────

def draw_rectangle(x, y, rect_length, rect_width, fill_color):
    """
    Draw a filled rectangle centered at (x, y).

    Parameters
    ----------
    x, y        : center position  (0,0 = canvas center)
    rect_length : height of the rectangle
    rect_width  : width  of the rectangle
    fill_color  : color name ("red") or hex string ("#FF0000")
    """
    setheading(0)
    penup()
    _goto(x - rect_width / 2, y - rect_length / 2)
    color(fill_color)
    begin_fill()
    pendown()
    for _ in range(2):
        forward(rect_width)
        left(90)
        forward(rect_length)
        left(90)
    end_fill()
    penup()
    setheading(0)


def draw_square(x, y, fill_color, size):
    """
    Draw a filled square centered at (x, y).

    Parameters
    ----------
    x, y       : center position
    fill_color : color name or hex string
    size       : side length in pixels
    """
    draw_rectangle(x, y, size, size, fill_color)


def draw_triangle(x, y, filled, fill_color, size):
    """
    Draw an upward-pointing equilateral triangle centered at (x, y).

    Parameters
    ----------
    x, y       : center position
    filled     : True → solid fill  |  False → outline only
    fill_color : color name or hex string
    size       : side length in pixels
    """
    h = math.sin(math.radians(60)) * size
    setheading(0)
    penup()
    color(fill_color)
    _goto(x - size / 2, y - h / 2)
    pendown()
    if filled:
        begin_fill()
    forward(size)
    left(120)
    forward(size)
    left(120)
    forward(size)
    left(120)
    if filled:
        end_fill()
    penup()
    setheading(0)


def draw_circle(x, y, fill_color, radius):
    """
    Draw a filled circle centered at (x, y).

    Parameters
    ----------
    x, y       : center position
    fill_color : color name or hex string
    radius     : circle radius in pixels
    """
    penup()
    setheading(0)
    _goto(x, y - radius)
    color(fill_color)
    begin_fill()
    pendown()
    circle(radius)
    end_fill()
    penup()


# ── LANDSCAPE ELEMENTS ───────────────────────────────────────────────────────

def draw_sky():
    """Fill the entire canvas with a sky-blue background."""
    setheading(0)
    penup()
    _goto(-CANVAS_WIDTH // 2, -CANVAS_HEIGHT // 2)
    color("#87CEEB")
    begin_fill()
    pendown()
    for _ in range(2):
        forward(CANVAS_WIDTH)
        left(90)
        forward(CANVAS_HEIGHT)
        left(90)
    end_fill()
    penup()
    setheading(0)


def draw_grass():
    """Draw a green grass strip across the lower portion of the canvas."""
    draw_rectangle(0, -CANVAS_HEIGHT // 4, CANVAS_HEIGHT // 2, CANVAS_WIDTH, "#228B22")


def draw_sun(x, y, radius):
    """
    Draw an orange sunburst at (x, y).

    Parameters
    ----------
    x, y   : center position
    radius : length of each sun ray in pixels
    """
    setheading(0)
    penup()
    _goto(x, y)
    color("orange")
    pendown()
    for _ in range(80):
        forward(radius)
        left(170)
    penup()


def draw_cloud(x, y, size):
    """
    Draw a white fluffy cloud near (x, y).

    Parameters
    ----------
    x, y : approximate center-left of the cloud
    size : radius of each cloud puff in pixels
    """
    penup()
    setheading(0)
    color("white")
    for ox, oy in [(0, 0), (size, int(size * 0.35)), (-size, int(size * 0.35))]:
        _goto(x + ox, y + oy - size)
        begin_fill()
        pendown()
        circle(size)
        end_fill()
        penup()


def draw_bird(x, y, bird_color, size):
    """
    Draw a V-shaped bird at (x, y).

    Parameters
    ----------
    x, y       : wing-center position
    bird_color : color name or hex string
    size       : wingspan length in pixels
    """
    setheading(0)
    color(bird_color)
    penup()
    _goto(x, y)
    pendown()
    left(45);  forward(size);  right(90);  forward(size / 2)
    penup()
    _goto(x, y)
    setheading(0)
    pendown()
    left(135); forward(size);  left(90);   forward(size / 2)
    penup()
    setheading(0)


def draw_tree(x, y, size):
    """
    Draw a single layered pine tree centered at (x, y).

    Parameters
    ----------
    x, y : base center position
    size : controls overall tree size
    """
    tri_height = math.sin(math.radians(60)) * size
    draw_triangle(x, y,                       True, "#2E8B57", size)
    draw_triangle(x, y - (60  / 200) * size,  True, "#228B22", size)
    draw_triangle(x, y - (120 / 200) * size,  True, "#006400", size)
    trunk_h = (80 / 200) * size
    trunk_w = (30 / 200) * size
    draw_rectangle(x, y - tri_height - trunk_h / 2, trunk_h, trunk_w, "#8B4513")


def draw_trees(x, y, num_trees, size):
    """
    Draw a row of pine trees starting at (x, y).
    Trees at higher y values appear smaller (depth effect).

    Parameters
    ----------
    x, y      : starting position of the leftmost tree
    num_trees : how many trees to draw
    size      : base tree size
    """
    scaling = 0.1
    for n in range(num_trees):
        jitter = random.randint(-2, 2)
        if y < 0:
            size_val = size + abs(scaling * y)
        elif size - scaling * y <= 0:
            size_val = 10
        else:
            size_val = abs(size - scaling * y)
        draw_tree(x + n * (160 / 200) * size,
                  y + jitter * (20  / 200) * size,
                  size_val)


def draw_house(x, y, house_color, size):
    """
    Draw a house (square body + triangle roof) centered at (x, y).

    Parameters
    ----------
    x, y        : center of the square body
    house_color : color name or hex string
    size        : side length of the square body in pixels
    """
    draw_square(x, y, house_color, size)
    roof_y = y + size / 2 + (math.sin(math.radians(60)) * size) / 2
    draw_triangle(x, roof_y, True, house_color, size)


def draw_houses(x, y, num_houses, size):
    """
    Draw a row of houses starting at (x, y) with random colors.

    Parameters
    ----------
    x, y       : starting position of the leftmost house
    num_houses : how many houses to draw
    size       : side length of each house body in pixels
    """
    for n in range(num_houses):
        jitter = random.randint(-8, 8)
        draw_house(x + n * 1.3 * size, y + jitter, get_random_color(), size)


def draw_rock(x, y, rock_color, size):
    """
    Draw an irregular rock shape at (x, y).

    Parameters
    ----------
    x, y       : approximate center position
    rock_color : color name or hex string
    size       : controls overall rock size
    """
    setheading(0)
    penup()
    _goto(x - size / 2, y - size / 2)
    color(rock_color)
    begin_fill()
    pendown()
    for angle, dist in [(10, size),        (15, 0.75 * size), (85, size),
                        (15, size * 0.333), (10, size * 0.5),  (45, size),
                        (35, size / 1.3),   (50, size),        (15, size / 2),
                        (70, size / 1.2)]:
        left(angle)
        forward(dist)
    end_fill()
    penup()
    setheading(0)


def get_random_color():
    """Return a random pastel house color."""
    return random.choice([
        "#E8A87C", "#85DCBA", "#F7D794", "#778CA3",
        "#CF6A87", "#AED6F1", "#A9CCE3", "#F0B27A"
    ])

print("✅ All helper functions loaded! Scroll down to draw your landscape.")
```

## 🎨 Your Landscape — Edit & Run

**Edit `draw_landscape()` below to build your scene.** Run the cell whenever you want to see your drawing.

| What to change | How |
|---|---|
| Move an object | Change the `x` and `y` values |
| Make it bigger / smaller | Change `size` or `radius` |
| Change a color | Use a name like `"red"` or a hex code like `"#FF6347"` |
| Add more objects | Call any helper function again with new values |
| Remove something | Delete or comment out (`#`) that line |

**Coordinate reminder** — `(0, 0)` is the **canvas center** · x → right (negative = left) · y → up (negative = down) · canvas ≈ ±400 wide, ±250 tall

**Required: keep at least one of each →** `draw_sky` `draw_grass` `draw_sun` `draw_cloud` `draw_bird` `draw_tree` `draw_house` `draw_rock` `draw_circle` `draw_square` `draw_triangle` `draw_rectangle`

```python
# ============================================================
#  CELL 3 — Edit draw_landscape() to create YOUR scene!
# ============================================================

def draw_landscape():
    init_scene()   # ← always keep this line first

    # ── BACKGROUND (draw these first) ───────────────────────
    draw_sky()
    draw_grass()

    # ── SUN ─────────────────────────────────────────────────
    #  draw_sun(x, y, radius)
    draw_sun(0, 190, 75)

    # ── CLOUDS ──────────────────────────────────────────────
    #  draw_cloud(x, y, size)
    draw_cloud(-220, 165, 25)
    draw_cloud( 160, 185, 20)
    draw_cloud( 330, 170, 22)

    # ── BIRDS ───────────────────────────────────────────────
    #  draw_bird(x, y, color, size)
    draw_bird(-170, 135, "black", 13)
    draw_bird( -20, 150, "black", 10)
    draw_bird( 210, 140, "black", 11)

    # ── TREES (back row — smaller because y is higher) ──────
    #  draw_trees(x, y, num_trees, size)
    draw_trees(-360, 30, 5, 65)

    # ── HOUSES ──────────────────────────────────────────────
    #  draw_houses(x, y, num_houses, size)
    draw_houses(-130, -40, 4, 42)

    # ── TREES (front row — larger because y is lower) ────────
    draw_trees(210, -70, 3, 85)

    # ── ROCKS ───────────────────────────────────────────────
    #  draw_rock(x, y, color, size)
    draw_rock(-230, -165, "#808080", 18)
    draw_rock( 285, -170, "#696969", 15)

    # ── SHAPES ──────────────────────────────────────────────
    #  draw_circle(x, y, color, radius)
    draw_circle(-340, -165, "#9B59B6", 18)

    #  draw_square(x, y, color, size)
    draw_square(350, 75, "#3498DB", 32)

    #  draw_triangle(x, y, filled, color, size)
    draw_triangle(335, -85, True, "#E74C3C", 42)

    #  draw_rectangle(x, y, rect_length, rect_width, color)
    draw_rectangle(-350, 45, 55, 22, "#F39C12")


# ── Run your scene ───────────────────────────────────────────
draw_landscape()
```
